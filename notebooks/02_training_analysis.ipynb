Analyzes loss curves, gradient norms, EMA parameter divergence, and learning rate schedules.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.evaluation.metrics import load_metrics_from_csv

df = load_metrics_from_csv('results/metrics/results.csv')
if not df.empty:
    print(df[['experiment_id', 'model', 'sde', 'training_time', 'fid', 'inception_score']])
else:
    print('No training metrics recorded yet.')
```

```python
if not df.empty and 'training_time' in df.columns:
    plt.figure(figsize=(8, 4.5))
    sns.barplot(data=df, x='model', y='training_time', hue='sde')
    plt.title('Training Duration by Architecture and SDE Formulation')
    plt.ylabel('Seconds')
    plt.show()
```
