# 03. Generative Quality Evaluation: FID & Inception Score

In-depth analysis of sample distribution fidelity (Fréchet Inception Distance) and semantic diversity (Inception Score).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.evaluation.metrics import load_metrics_from_csv

df = load_metrics_from_csv('results/metrics/results.csv')
if not df.empty:
    plt.figure(figsize=(9, 5))
    sns.scatterplot(data=df, x='fid', y='inception_score', hue='model', style='sde', s=120)
    plt.title('FID vs Inception Score Pareto Frontier')
    plt.xlabel('Fréchet Inception Distance (FID, Lower is Better)')
    plt.ylabel('Inception Score (IS, Higher is Better)')
    plt.grid(True)
    plt.show()
```
